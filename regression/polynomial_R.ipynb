# Polynomial Regression

Definition:
Used when the relationship between X and Y is non-linear (curved).

Formula:
y = b₀ + b₁x + b₂x² + ... + bₙxⁿ

Use Case:
When a straight line cannot fit the data properly.

Degree 1 → Linear Regression
Degree 2 → Quadratic Curve
Degree 3 → Cubic Curve

Advantage:
Captures non-linear relationships.

Disadvantage:
High degree can lead to overfitting.

```python
import operator
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error,r2_score
from sklearn.preprocessing import PolynomialFeatures
```

```python
np.random.seed(0)
x = 2 - 3*np.random.normal(0,1,20)
y = x - 2*(x**2) + 0.5*(x**3) + np.random.normal(-3,3,20)
```

```python
x
```

```python
y
```

```python
x = x[:,np.newaxis]
x
```

```python
y = y[:,np.newaxis]
y
```

```python
model = LinearRegression()
model.fit(x,y)
y_pred = model.predict(x)
```

```python
plt.scatter(x,y,s = 10)

sort_axis = operator.itemgetter(0)
sorted_zip = sorted(zip(x,y_pred),key = sort_axis)
x_sorted,y_pred_sorted = zip(*sorted_zip) # zip(*sorted_zip) unzips the sorted (x, y_pred) pairs back into separate x and y_pred arrays so they can be plotted.

plt.plot(x_sorted,y_pred_sorted,'red')
```

```python
RMSE = np.sqrt(mean_squared_error(y,y_pred))
r2 = r2_score(y,y_pred)
print("rmse is:",RMSE)
print("r2 score is:",r2*100)
```

### Polynomial Regression degree = 2

```python
np.random.seed(0)
x = 2 - 3*np.random.normal(0,1,20)
y = x - 2*(x**2) + 0.5*(x**3) + np.random.normal(-3,3,20)
```

```python
x = x[:,np.newaxis]
x
```

```python
Polynomial_features = PolynomialFeatures(degree = 2)
x_poly = Polynomial_features.fit_transform(x)
```

```python
model = LinearRegression()
model.fit(x_poly,y)
y_pred = model.predict(x_poly)
```

```python
plt.scatter(x,y,s = 10)

sort_axis = operator.itemgetter(0)
sorted_zip = sorted(zip(x,y_pred),key = sort_axis)
x_sorted,y_pred_sorted = zip(*sorted_zip) # zip(*sorted_zip) unzips the sorted (x, y_pred) pairs back into separate x and y_pred arrays so they can be plotted.

plt.plot(x_sorted,y_pred_sorted,'purple')
```

```python
RMSE = np.sqrt(mean_squared_error(y,y_pred))
r2 = r2_score(y,y_pred)
print("rmse is:",RMSE)
print("r2 score is:",r2*100)
```
