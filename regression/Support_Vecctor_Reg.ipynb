# Support Vector Regression (SVR)

## Overview

Support Vector Regression (SVR) is a regression algorithm based on Support Vector Machines (SVM). It predicts continuous values by finding the best-fit hyperplane within a specified error margin (ε).

---

## Key Idea

SVR does not try to minimize the error for every data point. Instead, it fits a hyperplane such that most points lie within an acceptable error range.

```text
y = wx + b
```

The fitted line is called the **Hyperplane**.

---

## Support Vectors

Support Vectors are the data points closest to the margin. They determine the position of the hyperplane.

---

## Kernel Trick

For non-linear data, SVR uses kernels to transform data into a higher-dimensional space.

Common kernels:
- Linear
- Polynomial
- RBF (Most Common)
- Sigmoid

---

## Important Parameters

- **C** → Controls model flexibility and error penalty.
- **ε (epsilon)** → Width of the error margin.
- **Kernel** → Determines the transformation method.

---

## Advantages

- Handles linear and non-linear data.
- Effective on small to medium-sized datasets.

---

## Limitations

- Sensitive to feature scaling.
- Computationally expensive for large datasets.

---

## One-Line Definition

> SVR predicts continuous values by fitting an optimal hyperplane within a specified error margin while using support vectors to define the model.

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
````

````python
dataset = pd.read_csv(r"C:\Users\DELL\Desktop\course_prac\ml\regression\Student_Marks.csv")
dataset.head()
````

````python
x = dataset.iloc[:,0].values
y = dataset.iloc[:,1].values

x = np.array(x).reshape(-1,1)
y = np.array(y).reshape(-1,1)
````

````python
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.2,random_state = 42)
````

````python
from sklearn.svm import SVR

regressor = SVR(kernel = 'rbf')
regressor.fit(x_train,y_train)
````

````python
y_pred = regressor.predict(x_test)
y_pred
````

````python
df = pd.DataFrame({'real values':y_test.reshape(-1),'predicted values':y_pred})
df
````

````python
x_grid = np.arange(min(x),max(x),0.1)
x_grid = x_grid.reshape(len(x_grid),1)

plt.scatter(x_test,y_test.reshape(-1),color = 'blue')
plt.scatter(x_test,y_pred,color = 'red')

plt.title('SVR')
plt.xlabel('hours of study')
plt.ylabel('marks')
plt.plot()
````
