**Classification** can be performed on structured or unstructured data. Classification is a technique where we categorize data into a given number of classes. The main goal of a classification problem is to identify the category/class to which a new data will fall under.

Few of the terminologies encountered in machine learning – classification:

**Classifier:** An algorithm that maps the input data to a specific category.

**Classification model:** A classification model tries to draw some conclusion from the input values given for training. It will predict the class labels/categories for the new data.

**Feature:** A feature is an individual measurable property of a phenomenon being observed.

**Binary Classification:** Classification task with two possible outcomes. Eg: Gender classification (Male / Female)

**Multi-class classification:** Classification with more than two classes. In multi class classification each sample is assigned to one and only one target label. Eg: An animal can be cat or dog but not both at the same time

**Multi-label classification:** Classification task where each sample is mapped to a set of target labels (more than one class). Eg: A news article can be about sports, a person, and location at the same time.

Performance Metrics:

**Model accuracy** is a machine learning model performance metric that is defined as the ratio of true positives and true negatives to all positive and negative observations. In other words, accuracy tells us how often we can expect our machine learning model will correctly predict an outcome out of the total number of times it made predictions.The accuracy rate is great but it doesn’t tell us anything about the errors our machine learning models make on new data we haven’t seen before.

Mathematically, it represents the ratio of the sum of true positive and true negatives out of all the predictions.

Accuracy Score = (TP + TN)/ (TP + FN + TN + FP)

**Precision:** Model precision score represents the model’s ability to correctly predict the positives out of all the positive prediction it made. Precision score is a useful measure of success of prediction when the classes are very imbalanced. Mathematically, it represents the ratio of true positive to the sum of true positive and false positive.

Precision Score = TP / (TP+ FP)

**Recall:** Model recall score represents the model’s ability to correctly predict the positives out of actual positives. This is unlike precision which measures how many predictions made by models are actually positive out of all positive predictions made.In other words, it measures how good our machine learning model is at identifying all actual positives out of all positives that exist within a dataset. The higher the recall score, the better machine learning model is at identifying both positive and negative examples. Recall score is a useful measure of success of prediction when the classes are very imbalanced.  Mathematically, it represents the ratio of true positive to the sum of true positive and false negative.

Recall Score = TP / (TP + FN)

**Model F1 score** represents the model score as a function of precision and recall score. F-score is a machine learning model performance metric that gives equal weight to both the Precision and Recall for measuring its performance in terms of accuracy, making it an alternative to Accuracy metrics (it doesn’t require us to know the total number of observations). It’s often used as a single value that provides high-level information about the model’s output quality. This is useful measure of the model in the scenarios where one tries to optimize either of precision or recall score and as a result, the model performance suffers.Mathematically, it can be represented as harmonic mean of precision and recall score.

F1 Score = 2* Precision Score * Recall Score/ (Precision Score + Recall Score/)

### Difference Between Precision, Recall, and F1 Score
Precision, recall, and F1 score are important metrics for evaluating classification models, especially when dealing with imbalanced datasets.

**Precision** measures how many of the predicted positive cases were actually positive. It answers: "Of all the items labeled as positive by the model, how many are truly positive?"
- Formula: Precision = TP / (TP + FP)
- High precision means that when the model predicts positive, it is usually correct.

**Recall** (also called Sensitivity or True Positive Rate) measures how many of the actual positive cases were correctly identified by the model. It answers: "Of all the actual positive items, how many did the model correctly identify?"
- Formula: Recall = TP / (TP + FN)
- High recall means the model finds most of the positive cases.

**F1 Score** is the harmonic mean of precision and recall. It balances both metrics, providing a single score that considers both false positives and false negatives.
- Formula: F1 Score = 2 * (Precision * Recall) / (Precision + Recall)
- Useful when you want a balance between precision and recall, especially if you have uneven class distribution.

**Summary Table:**
| Metric    | What it measures | Formula                |
|-----------|------------------|------------------------|
| Precision | Correctness of positive predictions | TP / (TP + FP) |
| Recall    | Coverage of actual positives        | TP / (TP + FN) |
| F1 Score  | Balance between precision and recall| 2 * (Precision * Recall) / (Precision + Recall) |

**Example:**
- If your model has high precision but low recall, it is very accurate when it predicts positive, but it misses many actual positives.
- If your model has high recall but low precision, it finds most positives, but also includes many false positives.
- F1 score helps you find a balance between the two.

# Logistic Regression – Quick Notes

## What is Logistic Regression?
- A **classification algorithm** used to predict categorical outcomes.
- Gives output as a **probability (0 to 1)**.
- Uses the **Sigmoid Function** to convert predictions into probabilities.

### Examples
- Spam / Not Spam
- Fraud / Not Fraud
- Yes / No
- 0 / 1
- Tumor Benign / Malignant

---

## When to Use Logistic Regression?
✅ Target variable is categorical (binary or multiclass)

✅ Need probability scores

✅ Data is approximately linearly separable

✅ Need model interpretability (feature impact)

---

## Types of Logistic Regression

### 1. Binary Logistic Regression
- Two classes only
- Example: Yes/No, Pass/Fail

### 2. Multinomial Logistic Regression
- More than two classes
- Example: Cat/Dog/Sheep

### 3. Ordinal Logistic Regression
- Ordered categories
- Example: Low, Medium, High

---

## Logistic Regression vs Linear Regression

| Linear Regression | Logistic Regression |
|------------------|--------------------|
| Predicts continuous values | Predicts categories |
| Output can be any value | Output between 0 and 1 |
| Uses Linear Function | Uses Sigmoid Function |

---

## Can Linear Regression Be Used for Classification?
**Yes, but not recommended.**

Reason:
- Apply a threshold (e.g., 0.5)
- Values > 0.5 → Class 1
- Values ≤ 0.5 → Class 0

Problems:
- Predictions may be < 0 or > 1
- Poor classification performance
- Higher error

---

## Sigmoid Function

Converts any real value into a probability between **0 and 1**.

### Formula

\[
\sigma(z) = \frac{1}{1 + e^{-z}}
\]

### Key Points
- Output range: **0 to 1**
- Output ≈ 0 → Class 0
- Output ≈ 1 → Class 1
- Output = 0.5 → Decision boundary

---

## Cost Function (Log Loss)

Purpose:
- Measures prediction error.
- Goal is to **minimize the cost**.

### Objective

\[
\text{Minimize } J(\theta)
\]

Lower cost = Better model.

---

## Gradient Descent

Used to find optimal model parameters.

### Steps
1. Initialize weights
2. Calculate cost
3. Update weights
4. Repeat until cost is minimized

### Update Rule

\[
\theta = \theta - \alpha \frac{\partial J}{\partial \theta}
\]

where:
- θ = parameter/weight
- α = learning rate

---

## Advantages
✔ Simple and fast

✔ Easy to interpret

✔ Produces probability scores

✔ Works well for binary classification

✔ Less prone to overfitting than complex models

---

## Disadvantages
✘ Assumes linear relationship

✘ Sensitive to outliers

✘ May struggle with complex non-linear patterns

✘ Performance decreases when classes overlap heavily

---

# Naive Bayes – Quick Notes

## What is Naive Bayes?
- A **supervised machine learning classification algorithm**.
- Based on **Bayes' Theorem**.
- Assumes all input features are **independent** of each other.
- Predicts the class with the **highest probability**.

---

## Bayes' Theorem

\[
P(A|B) = \frac{P(B|A) \times P(A)}{P(B)}
\]

Where:
- **P(A)** → Prior Probability
- **P(B)** → Evidence
- **P(B|A)** → Likelihood
- **P(A|B)** → Posterior Probability

---

## Why is it called "Naive"?
Because it assumes that all features are independent, which is often not true in real-world data.

---

## Types of Naive Bayes

### 1. Gaussian Naive Bayes
- Used for **continuous numerical data**.
- Assumes data follows a **Normal (Gaussian) Distribution**.

### 2. Multinomial Naive Bayes
- Used for **text classification**.
- Works with word frequencies/counts.

### 3. Bernoulli Naive Bayes
- Used for **binary features (0/1, Yes/No)**.

---

## Advantages
✔ Fast training and prediction

✔ Works well on small datasets

✔ Handles multiclass classification

✔ Excellent for text classification

---

## Disadvantages
✘ Assumes feature independence

✘ Not suitable when features are highly correlated

✘ Probability estimates may not be very accurate

---

## Applications
- Spam Email Detection
- Sentiment Analysis
- Text Classification
- Recommendation Systems
- Real-time Prediction

---

# Stochastic Gradient Descent (SGD) – Quick Notes

## What is Gradient Descent?
- An optimization algorithm used to minimize the **cost/loss function**.
- Finds the best model parameters (weights).
- Moves step-by-step towards the minimum error.

### Update Rule

\[
\theta = \theta - \alpha \frac{\partial J}{\partial \theta}
\]
θnew = θold - α × (∂J/∂θ)

Where:
- θ = Weight/Parameter
- α = Learning Rate
- J = Cost Function

---

## What is Stochastic Gradient Descent (SGD)?
- A faster version of Gradient Descent.
- Instead of using the entire dataset, it updates weights using **one random data point** (or a small batch).
- Reduces computation time on large datasets.

---

## Why Use SGD?
✔ Faster training

✔ Suitable for large datasets

✔ Requires less memory

✔ Widely used in Deep Learning

---

## Types

### 1. Batch Gradient Descent
- Uses entire dataset for each update.
- Accurate but slow.

### 2. Stochastic Gradient Descent (SGD)
- Uses one random sample for each update.
- Fast but noisy.

### 3. Mini-Batch Gradient Descent
- Uses a small batch of samples.
- Most commonly used in practice.

---

## Learning Rate (α)
- Controls step size during optimization.

### High Learning Rate
❌ May overshoot minimum

### Low Learning Rate
❌ Slow convergence

### Optimal Learning Rate
✔ Faster and stable convergence

---

## Advantages
✔ Faster than Gradient Descent

✔ Works well with huge datasets

✔ Less memory usage

✔ Helps train Neural Networks efficiently

---

## Disadvantages
✘ Updates can be noisy

✘ May not reach exact minimum

✘ Sensitive to learning rate selection

---

## Popular SGD-Based Optimizers
- Adam
- Adagrad
- Adadelta
- RMSProp

---

# K-Nearest Neighbors (KNN)

## What is KNN?
- Supervised Learning Algorithm
- Used for Classification and Regression
- Predicts using the K nearest data points (neighbors)

---

## Characteristics of KNN

### Instance-Based Learning
- Uses training data directly for prediction.
- No model is built.

### Lazy Learning
- Stores training data.
- Learning happens only during prediction.

### Non-Parametric
- No assumption about data distribution.

---

## How KNN Works?

1. Choose K value
2. Calculate distance from new point to all training points
3. Select K nearest neighbors
4. Classification → Majority Vote
5. Regression → Mean/Average of neighbors

---

## Distance Metric

### Euclidean Distance

\[
d = \sqrt{\sum_{i=1}^{n}(x_i-y_i)^2}
\]

Most commonly used distance measure.

Other distances:
- Manhattan Distance
- Minkowski Distance
- Hamming Distance

---

## Choosing K Value

### Small K
- High Variance
- Overfitting

Example:
```text
K = 1
```

### Large K
- High Bias
- Underfitting

### Optimal K
- Value where Test Error is minimum.
- Usually found using Cross Validation.

### Rule
For binary classification:

```text
Use Odd K (3,5,7...)
```

to avoid ties.

---

## Classification vs Regression

### Classification
- Majority class among K neighbors becomes prediction.

### Regression
- Mean/Average of K neighbors becomes prediction.

---

## Advantages
✔ Simple and easy to understand

✔ No training phase

✔ Works for Classification and Regression

✔ Robust to noisy data

---

## Disadvantages
✘ Slow on large datasets

✘ Computationally expensive

✘ Sensitive to feature scaling

✘ Choosing K can be difficult

---

## Important Point

Always scale features before KNN:

```python
StandardScaler()
```

or

```python
MinMaxScaler()
```

because KNN is distance-based.

---

## Interview One-Liner

"KNN is a supervised, instance-based, lazy learning algorithm that predicts a class or value using the K nearest neighbors based on a distance metric such as Euclidean distance."

````python
#Reading Dataset from csv file
df = pd.read_table(r"C:\Users\DELL\Desktop\course_prac\ml\Classification\fruit_data_with_colors.txt")
df.head()
````

````python
#Input and Output Variables
feature_names = ['mass', 'width', 'height', 'color_score']
X = df[feature_names]
y = df['fruit_label']
y.unique() #Multiclass
````

````python
#Scaling X and y
scaler = MinMaxScaler()
scaled_X = scaler.fit_transform(X)

#Splitting dataset into Train Test 70/30
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.30, random_state=1, stratify = y)
print(X_train.shape, X_test.shape, y_train.shape, y_test.shape)
````

# Support Vector Machine (SVM)

## What is SVM?
- Supervised Machine Learning Algorithm
- Mainly used for Classification (can also be used for Regression)
- Finds the best decision boundary to separate classes

---

## Important Terms

### Hyperplane
- Decision boundary that separates classes.
- Best hyperplane = Maximum Margin Hyperplane.

### Support Vectors
- Data points closest to the hyperplane.
- They determine the position of the hyperplane.

### Margin
- Distance between support vectors and hyperplane.
- SVM aims to maximize this margin.

---

## Types of SVM

### Linear SVM
- Used when data is linearly separable.
- Uses a straight line (2D) or plane (higher dimensions).

### Non-Linear SVM
- Used when data cannot be separated by a straight line.
- Uses the Kernel Trick.

---

## Kernel Trick
- Converts low-dimensional data into higher-dimensional space.
- Helps separate non-linear data.

Common Kernels:
- Linear
- Polynomial
- RBF (Gaussian)
- Sigmoid

---

## How SVM Works?

1. Find possible decision boundaries.
2. Identify support vectors.
3. Calculate margins.
4. Select hyperplane with maximum margin.
5. Use it for classification.

---

## Advantages
✔ Works well for high-dimensional data

✔ Effective when classes are clearly separated

✔ Memory efficient (uses support vectors)

✔ Works for linear and non-linear data

---

## Disadvantages
✘ Slow on very large datasets

✘ Sensitive to noisy/overlapping data

✘ More difficult to interpret

---

## Applications
- Face Detection
- Image Classification
- Text Classification
- Spam Detection

---
