# Bagging vs Boosting

Bagging and Boosting are **Ensemble Learning Techniques**. Ensemble learning combines multiple machine learning models to produce a better, more accurate, and more robust final model than a single model.

---

# Ensemble Learning

Instead of relying on one model,

```
Model 1
Model 2
Model 3
Model 4
      ↓
Combine Predictions
      ↓
Final Prediction
```

Advantages:
- Higher Accuracy
- Better Generalization
- Reduced Overfitting
- More Stable Predictions

Ensemble learning is mainly divided into:

1. Bagging (Bootstrap Aggregating)
2. Boosting

---

# Bagging (Bootstrap Aggregating)

## Definition

Bagging trains multiple models **independently and in parallel** using different random subsets of the training data.

Each model gives its own prediction.

Final prediction is:

- **Majority Vote** (Classification)
- **Average** (Regression)

Bagging mainly helps to **reduce Variance**.

---

# Why is it called Bootstrap Aggregating?

Bagging consists of two words:

## Bootstrap

Random sampling **with replacement**.

Example:

Original Data

```
A B C D E
```

Bootstrap Sample

```
A C C E B
```

Notice:

- C appears twice
- D is missing

This is called **Sampling With Replacement**.

---

## Aggregating

After all models predict,

Combine their predictions.

Classification

```
Tree1 → 0
Tree2 → 1
Tree3 → 0
Tree4 → 0
Tree5 → 1

Majority = 0
```

Regression

```
200
220
250
210
240

Average = 224
```

---

# Working of Bagging

Suppose we have

```
10000 rows
20 features
```

We decide to build

```
10 Decision Trees
```

Each tree randomly selects

```
Rows → 2000

Columns → 12
```

Example

```
Tree 1

Rows:
1,15,22,100,...

Columns:
Age
Salary
Experience
Loan
```

Tree 2

```
Different Rows

Different Columns
```

Every tree is different.

Finally,

```
Tree1
Tree2
Tree3
Tree4
...
Tree10

↓

Voting / Averaging

↓

Final Prediction
```

---

# Parallel Nature

Bagging is **Parallel**.

```
            Dataset
                |
-----------------------------------------
|      |      |      |      |      |
DT1   DT2    DT3    DT4    DT5

All train simultaneously
```

---

# Advantages of Bagging

- Reduces Variance
- Reduces Overfitting
- Stable Predictions
- Works well for noisy datasets
- Parallel training makes it faster

---

# Disadvantages

- Training many models increases memory usage
- Less interpretable
- Slightly slower during prediction

---

# Algorithms used in Bagging

Bagging can technically be applied to **any base estimator**.

Examples:

### Classification

- Decision Tree ⭐ (Most Common)
- Logistic Regression
- KNN
- SVM
- Naive Bayes

### Regression

- Decision Tree Regressor ⭐
- Linear Regression
- KNN Regressor
- SVR

However,

**Decision Trees are almost always used because they have high variance.**

---

# Random Forest

Random Forest is the most popular Bagging algorithm.

It consists of

```
Multiple Decision Trees

+

Bootstrap Sampling

+

Random Feature Selection
```

Final prediction

Classification

```
Majority Vote
```

Regression

```
Average
```

---

## Random Forest Hyperparameters

### n_estimators

Number of Decision Trees.

```
n_estimators=100

↓

100 Trees
```

---

### max_depth

Maximum depth of every tree.

Higher

↓

More complex tree

Lower

↓

Simpler tree

---

### max_features

Number of features each tree can consider while splitting.

# Boosting

## Definition

Boosting trains models **Sequentially**.

Each new model learns from the mistakes made by the previous model.

Main objective

**Reduce Bias**

---

# Working of Boosting

Suppose

Model 1 predicts

```
90 Correct

10 Wrong
```

Now

Model 2 focuses mainly on

```
Those 10 wrong samples.
```

Again

Model 2 predicts

Only

```
3 mistakes
```

Model 3 focuses on

```
Those 3 mistakes.
```

This process continues.

---

# Sequential Nature

Boosting is Sequential.

```
Dataset

↓

Model 1

↓

Errors

↓

Model 2

↓

Errors

↓

Model 3

↓

Errors

↓

Final Model
```

Models cannot be trained simultaneously.

---

# Advantages

- Reduces Bias
- High Accuracy
- Excellent Performance
- Learns difficult patterns

---

# Disadvantages

- Training is slow
- Sensitive to noisy data
- Can overfit if not tuned properly

---

# Algorithms used in Boosting

Boosting can also use different weak learners, but in practice, **Decision Trees (Decision Stumps or Shallow Trees)** are almost always used.

Common Boosting algorithms:

- AdaBoost
- Gradient Boosting
- XGBoost ⭐
- LightGBM
- CatBoost

Theoretically, boosting can be built using other learners like Logistic Regression or SVM, but modern boosting libraries are specifically designed around decision trees.

---

# XGBoost

XGBoost stands for

**Extreme Gradient Boosting**

It is an optimized implementation of Gradient Boosting.

Features

- Regularization (L1 + L2)
- Parallel computation
- Faster training
- Handles Missing Values
- Built-in Cross Validation
- Early Stopping

---

# Why XGBoost is Powerful

Each tree tries to correct

```
Errors

↓

Residuals

↓

Gradient

↓

New Tree

↓

Better Prediction
```

It also controls overfitting using

- max_depth
- gamma
- lambda
- alpha
- learning_rate

---

# Bagging vs Boosting

| Feature | Bagging | Boosting |
|----------|----------|-----------|

# Memory Trick

## Bagging

```
Bootstrap
↓

Parallel

↓

Independent Models

↓

Majority Vote / Average

↓

Reduce Variance
```

---

## Boosting

```
Sequential

↓

Correct Previous Errors

↓

Weighted Learning

↓

Reduce Bias

↓

High Accuracy
```

# Interview Questions

### Q1. Why is Random Forest called Bagging?

Because every tree is trained using **Bootstrap Sampling** and the predictions are **Aggregated**.

---

### Q2. Why does Bagging reduce Variance?

Different models learn different subsets of data.

Combining them reduces fluctuations in predictions.

---

### Q3. Why does Boosting reduce Bias?

Each new model learns from the mistakes made by previous models, gradually improving the overall prediction.

---

### Q4. Which is faster?

Bagging, because models are trained in parallel.

---

### Q5. Which usually gives higher accuracy?

XGBoost (Boosting), when properly tuned.

---

### Q6. Which algorithm is most commonly used for Bagging and Boosting?

**Decision Trees** are the standard base learner for both techniques because they are easy to train, can capture complex patterns, and benefit the most from ensemble methods.
